```python
class DataAnalysis:
    def __init__(self, data, crimes):
        self.data = data
        self.crimes = crimes
        self.average_sentences = self.crime_builder(crimes)
        

    def crime_builder(self, crimes):
        average_sentences = {}
        
        for c in crimes:
            crime_specific_data = []
            for d in self.data:
                if d['charge'] == c:
                    crime_specific_data.append(d)
                else:
                    continue
            average_sentences[c] = crime_specific_data
        
        return average_sentences
    
    
```
