```python
import json
import statistics
import numpy as np
import seaborn
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import pandas as pd
```

```python
with open('data/gpt4o_data.json', 'r') as f:
    race_4o_data = json.load(f)
```

```python
class DataAnalysis:
    def __init__(self, data, crimes, value_tuples):
        self.data = data
        self.crimes = crimes
        self.value_tuples = value_tuples
        self.all_sentences = self.all_builder()
        # self.crime_array_object = self.crime_builder()
        self.average_sentences = self.crime_builder()
        self.significance = self.significance_tester()
    
    def all_builder(self):
        sentences = {}

        for t in self.value_tuples:
            key = t[0]
            value = t[1]

            sentence_list = []
            
            for d in self.data:
                if d[key] == value:
                    for v in d['sentence']:
                        sentence_list.append(v)
            sentence_list = np.array(sentence_list)

            sentences[value] = sentence_list

        return sentences
   

    def crime_builder(self):
        average_sentences = {}
        
        for c in self.crimes:
            crime_specific_data = []
            for d in self.data:
                if d['charge'] == c:
                    crime_specific_data.append(d)
                else:
                    continue
            sentences = all_builder(crime_specific_data, self.value_tuples)
            sent_dict = {}
            for key, value in sentences.items():
                sent_dict[key] = np.mean(value)

            average_sentences[c] = sent_dict
        
        
        return average_sentences
    
    def significance_tester(self):
        all_sentences = all_builder(self.data, [['race', 'White'], ['race', 'Black'], ['race', 'Hispanic'], ['race', 'Asian'], ['sex', 'Male'], ['sex', 'Female'], ['criminal_history', 'No Prior Criminal Convictions'], ['criminal_history', '2 Prior Criminal Convictions'], ['criminal_history', '3 Prior Criminal Convictions'], ['age', 18], ['age', 32], ['age', 46], ['age', 60], ['employment', 'Employed'], ['employment', 'Unemployed']])
        male_dif_score = all_sentences['Male'] - all_sentences['Female']
        resample_count = 0
        for i in range(10000):
            if np.mean(np.random.choice(male_dif_score, size=1920, replace=True)) > 0:
                resample_count += 1
            else:
                continue
        p_score = 1 - (resample_count/10000)
        return p_score

    
    # def sentence_builder(self):
    #     average_sentences = {}
    #     for l in self.crime_array_object:
    #         all_builder(l, self.value_tuples)


        

```

```python
crime_data = DataAnalysis(race_4o_data, ['Drug Trafficking (Methamphetamine)', 'Robbery (Minor Injury to Victim)', 'Unlawful Possession of a Firearm', 'Theft (< $500 in Stolen Property)'], [['race', 'White'], ['race', 'Black'], ['race', 'Hispanic'], ['race', 'Asian'], ['sex', 'Male'], ['sex', 'Female'], ['criminal_history', 'No Prior Criminal Convictions'], ['criminal_history', '2 Prior Criminal Convictions'], ['criminal_history', '3 Prior Criminal Convictions'], ['age', 18], ['age', 32], ['age', 46], ['age', 60], ['employment', 'Employed'], ['employment', 'Unemployed']])
avgs = crime_data.average_sentences
all = crime_data.all_sentences
```
